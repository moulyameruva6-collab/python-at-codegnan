```python
#Linear Search -- Time Complexity -0(n)
def LSearch(arr,val):
    for i in range(len(arr)):
        if arr[i] == val:
            return i
    return -1
#Driver Code    
arr=[10,2,45,3,7,24,90]
val=100
result = LSearch(arr,val)
print(result)
```

```python
def BSearch(arr,i,j,val):
    while i<<j:
        mid=i+j//2
        if arr[mid]==val:
            return mid
        elif arr[mid]>val: 
            j=mid-1
        else:
            i=mid+1
    return -1
    
# Drive Code
arr=[1,2,3,5,8,12,20,39,45,100,200]
val=5
i=0
j=len(arr)-1
result=BSearch(arr,i,j,val)
print(result)
```

```python
numbers = [10, 20, 30, 40, 50]

search = int(input("Enter value to search: "))

found = False

for i in range(len(numbers)):
    if numbers[i] == search:
        print("Value found at index:", i)
        found = True
        break

if not found:
    print("Value not found") 
```
