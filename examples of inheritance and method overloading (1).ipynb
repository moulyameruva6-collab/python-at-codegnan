```python
Definition:
Hierarchical inheritance is a type of inheritance where one parent (base) class is inherited by multiple child (derived) classes.

Real-Life Example
Parent Class: Vehicle
Child Classes: Car, Bike, Bus
```

```python
## 4. HEIRARICHAL INHERITANCE:
# Parent Class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def display_brand(self):
        print("Brand:", self.brand)


# Child Class 1
class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def display(self):
        print("Car Model:", self.model)


# Child Class 2
class Bike(Vehicle):
    def __init__(self, brand, cc):
        super().__init__(brand)
        self.cc = cc

    def display(self):
        print("Engine Capacity:", self.cc, )


# Child Class 3
class Bus(Vehicle):
    def __init__(self, brand, seats):
        super().__init__(brand)
        self.seats = seats

    def display(self):
        print("Number of Seats:", self.seats)


# Objects
c = Car("Toyota", "Fortuner")
b = Bike("Yamaha", 155)
bs = Bus("Volvo", 50)

print("Car Details")
c.display_brand()
c.display()

print()

print("Bike Details")
b.display_brand()
b.display()

print()

print("Bus Details")
bs.display_brand()
bs.display()
```

```python
Hybrid Inheritance in Python

Definition:
Hybrid inheritance is a combination of two or more types of inheritance (such as multiple inheritance + multilevel inheritance).

Note: Python supports hybrid inheritance because it supports multiple inheritance.
```

```python
## 5. HYBRID INHERITANCE:
# Parent Class
class Employee:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Employee:", self.name)


# Child Class 1
class Developer(Employee):
    def __init__(self, name):
        super().__init__(name)

    def coding(self):
        print(self.name, "writes code")


# Child Class 2
class Trainer(Employee):
    def __init__(self, name):
        super().__init__(name)

    def teach(self):
        print(self.name, "teaches students")


# Hybrid Child Class
class TechLead(Developer, Trainer):
    def __init__(self, name):
        super().__init__(name)

    def manage(self):
        print(self.name, "manages the team")


# Object
t = TechLead("Moulya")

t.show_name()
t.coding()
t.teach()
t.manage()
```

```python
## Arithmetic Operator Overloading:
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

    def __sub__(self, other):
        return self.value - other.value

    def __mul__(self, other):
        return self.value * other.value

    def __truediv__(self, other):
        return self.value / other.value

    def __floordiv__(self, other):
        return self.value // other.value

    def __mod__(self, other):
        return self.value % other.value

    def __pow__(self, other):
        return self.value ** other.value


n1 = Number(20)
n2 = Number(6)

print("Addition (+)       :", n1 + n2)
print("Subtraction (-)    :", n1 - n2)
print("Multiplication (*) :", n1 * n2)
print("Division (/)       :", n1 / n2)
print("Floor Division (//):", n1 // n2)
print("Modulus (%)        :", n1 % n2)
print("Power (**)         :", n1 ** n2)
```

```python
## Relational Operator Overloading:
class Student:
    def __init__(self, marks):
        self.marks = marks

    def __eq__(self, other):
        return self.marks == other.marks

    def __ne__(self, other):
        return self.marks != other.marks

    def __lt__(self, other):
        return self.marks < other.marks

    def __gt__(self, other):
        return self.marks > other.marks

    def __le__(self, other):
        return self.marks <= other.marks

    def __ge__(self, other):
        return self.marks >= other.marks


s1 = Student(85)
s2 = Student(75)

print("Equal (==)              :", s1 == s2)
print("Not Equal (!=)          :", s1 != s2)
print("Less Than (<)           :", s1 < s2)
print("Greater Than (>)        :", s1 > s2)
print("Less Than or Equal (<=) :", s1 <= s2)
print("Greater Than or Equal(>=):", s1 >= s2)
```

```python
class Demo:
    def __init__(self, name):
        self.name = name
        self.data = [10, 20, 30, 40]

    # String Representation
    def __str__(self):
        return f"Name: {self.name}"

    # Official Representation
    def __repr__(self):
        return f"Demo('{self.name}')"

    # Length
    def __len__(self):
        return len(self.data)

    # Membership
    def __contains__(self, item):
        return item in self.data

    # Indexing
    def __getitem__(self, index):
        return self.data[index]

    # Set Item
    def __setitem__(self, index, value):
        self.data[index] = value

    # Delete Item
    def __delitem__(self, index):
        del self.data[index]

    # Iterator
    def __iter__(self):
        return iter(self.data)

    # Callable
    def __call__(self):
        print("Object called like a function")

    # Boolean
    def __bool__(self):
        return len(self.data) > 0


# Object
obj = Demo("Moulya")

# __str__
print(obj)

# __repr__
print(repr(obj))

# __len__
print("Length:", len(obj))

# __contains__
print("20 in obj:", 20 in obj)

# __getitem__
print("First Element:", obj[0])

# __setitem__
obj[1] = 100
print("After Updating:", obj.data)

# __delitem__
del obj[2]
print("After Deleting:", obj.data)

# __iter__
print("Iterating Elements:")
for i in obj:
    print(i)

# __call__
obj()

# __bool__
print("Boolean Value:", bool(obj))
```
