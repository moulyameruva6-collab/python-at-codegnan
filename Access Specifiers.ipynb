```python
class Account:
    def __init__(self,name,balance,password):
        self.name=name # public
        self._balance=balance #protected
        self.__password=password #private
    
    def get_balance(self): #getter method
        return self._balance

    def update_password(self, new_password): #setter method
        self._password= new_password
        return "password updated"
    def get_password(self):
        return self.__password
        
    def withdraw(self,amount): #setter method
        if self._balance >= amount:
            self._balance <= amount
        return f"{amount} withdrawal successfully"
   
    def deposite(self,amount):  # setter method
        self._balance += amount
        return f"{amount} Deposite successfully and current balance is {self._balance}"
    def __total_users(self):  #private method
        return "total users :100"
        
user = Account("moulya",8000,"mouli@12")
print(user.name)
print(user.get_balance())
print(user.deposite(5000))
print(user.update_password("mouli@123"))
print(user.get_password())

#access the protected member syntax: obj_attributename
print(user._balance)
#access private member by using name mangling
#syntax: obj._classname__attributename
print(user._Account__total_users())
```
