```python
# Abstraction
from abc import ABC, abstractmethod


class Mobile(ABC):

    @abstractmethod
    def Ram(self):
        pass

    @abstractmethod
    def Rom(self):
        pass

    @abstractmethod
    def Battery(self):
        pass

    # Concrete method
    def about_device(self):
        return "This is Mobile"


class Oppo(Mobile):

    def __init__(self, model, Oppo_Ram, Oppo_Rom, Oppo_Battery):
        self.model = model
        self.Oppo_Ram = Oppo_Ram
        self.Oppo_Rom = Oppo_Rom
        self.Oppo_Battery = Oppo_Battery

    def Ram(self):
        return f"The Oppo {self.model} has {self.Oppo_Ram} GB RAM"

    def Rom(self):
        return f"The Oppo {self.model} has {self.Oppo_Rom} GB ROM"

    def Battery(self):
        return f"The Oppo {self.model} has {self.Oppo_Battery} mAh Battery"

    def Camera(self, front, back):
        return f"The Oppo {self.model} has {front} MP Front Camera + {back} MP Back Camera"


# Objects
reno16 = Oppo("Reno16", 16, 256, 5000)
reno15 = Oppo("Reno15", 8, 256, 5500)

print(reno16.about_device())
print(reno16.Ram())
print(reno16.Rom())
print(reno16.Battery())
print(reno16.Camera(32, 50))
```
