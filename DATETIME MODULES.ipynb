```python
from datetime import datetime, timedelta

curr_timestamp = datetime.now()   # current timestamp
print(curr_timestamp)

today = curr_timestamp.date()     # extracts date
curr_time = curr_timestamp.time() # extracts time

print(today)
print(curr_time)

print(today.year)
print(today.month)
print(today.day)

print(curr_time.hour, curr_time.minute, curr_time.second)

print(today.strftime("%d-%m-%Y"))
print(curr_time.strftime("%I:%M-%p"))
```

```python
import time
from IPython.display import clear_output
sec_counter = min_counter = hour_counter = 0
while True:
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter += 1
    if sec_counter == 60:
        min_counter +=1
        sec_counter +=0
    if min_counter == 60:
        hour_counter += 1
        min_counter +=1
    if hour_counter == 24:
        hour_counter == 0
    clear_output(wait=True)
```

```python
import time
from IPython.display import clear_output
sec_counter = min_counter = hour_counter = 0

while min_counter < 5:

    print(f"{hour_counter:02d}:{min_counter:02d}:{sec_counter:02d}")

    time.sleep(1)
    sec_counter += 1

    if sec_counter == 60:
        sec_counter = 0
        min_counter += 1

    if min_counter == 60:
        min_counter = 0
        hour_counter += 1

    clear_output(wait=True)

#print("Timer completed!")
```

```python
import time
from IPython.display import clear_output
minutes=int(input(" enter the stop watch timing:"))
sec_counter = min_counter = hour_counter = 0

while min_counter !=minutes:

    print(f"{hour_counter:02d}:{min_counter:02d}:{sec_counter:02d}")

    time.sleep(1)
    sec_counter += 1

    if sec_counter == 60:
        sec_counter = 0
        min_counter += 1

    if min_counter == 60:
        min_counter = 0
        hour_counter += 1

    clear_output(wait=True)

print("Timer completed!")
```
