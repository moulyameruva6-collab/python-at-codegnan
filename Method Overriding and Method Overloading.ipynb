```python
# POLYMORPHISM
##1. Method Overriding:
class Person:
    def  __init__(self,name,age):
        self.name=name
        self.age=age
    def details(self):
        return f" My name is {self.name} and I'm a{self.age} years old"

class Student(Person):
    def __init__(self,name,age,roll_no):
        self.roll_no=roll_no
        super().__init__(name,age)
    def details(self):
        return f"I'm a Student"
p=Person("Shannu",21)
s=Student("Moulya",20,101)
print(p.details())
print(s.details())        
```
