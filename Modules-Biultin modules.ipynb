```python
print(math.ceil(4.956)) # it returns the upper bound value
print(math.floor(4.25)) # it returns the lower bound value
print(math.factorial(5)) # it returns factorial of a number
print(math.gcd(10,5)) # it returns the GCD value
print(math.lcm(45,5,1)) # it returns the LCM value
print(math.log10(2))
print(math.log10(4))
print(math.log2(4))
print(math.pi)
```

```python
import random
print(random.random()) # this function returns the value in between 0-1
print(random.randint(1,10)) #  randint(a,b):this function generates random number between 1-10
print(random.choice(['A',1,2,3,4,5,6,7,8,9,'M','Q'])) # it returns the random items
print(random.choices(['A',1,2,3,4,5,6,7,8,9,'M','Q','K'],k=3))
arr = ['A',1,2,3,4,5,6,7,8,9,'M','Q']
random.shuffle(arr)
print(arr)


```

```python
#who will reach top in less steps:
import random
steps=rolls=0
while True:
    input("to roll dice for that click enter button")
    dice_val = random.randint(1,6)
    rolls += 1
    print("current  dice value:",dice_val)
    if steps + dice_val <= 100:
        steps += dice_val
    if steps == 100:
        print(f"your reached top in {rolls}rolls")
        break
    print(f"for{rolls} rolls your at {steps} position")
    
```
