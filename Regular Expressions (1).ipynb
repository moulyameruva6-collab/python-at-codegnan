```python
import re
```

```python
# Match(pattern,string)
string="The sky is blue and ocean is blue"
res=re.match("[A-Z]",string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern val
```

```python
# Match(pattern,string)
try:
    string="The sky is blue and ocean is blue"
    res=re.match("[z]+",string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val 
except Exception as e:
    print("There is no matched pattern:",e)
    
```

```python
#search (pattern,string):
#search 5 length word
string="The sky is blue and ocean is blue"
res=re.search("[a-zA-Z]{5}",string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern val
```

```python
import re
try:
    string="The sky is blue and oce456an is blue"
    res=re.findall("[A-Za-z]+",string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val 
except Exception as e:
    print("There is no matched pattern:",e)
   
```

```python
# validate mobile number
# All are digits , length of number is 10
# starts with 6,7,8,9 , remaining [0-9]

import re
try:
    mobile = input()
    
    res=re.fullmatch("^[6-9][0-9]{9}",mobile)
    if res:
        print(f"{mobile} is valid")
        print(res)
    else:
        print(f"{mobile} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
   
```

```python
#PAN CARD validation
#it allow upper case and digits ,length is 10
#first 5 are upper, next 4 are digits
#last letter ends,upper with [u-z]
import re
try:
    pancard=input()
    res=re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]$",pancard)
    if res:
        print(f"{pancard} is valid")
        print(res)
    else:
        print(f"{pancard} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
   
```

```python
## E-MAIL ID:
#small characters
#digits
#dot/special characters
import re
try:
    emailid=input()
    res=re.fullmatch("^[[a-z0-9][a-z0-9.]+\\@[a-z0-9.]+\\.[a-z]{2,3}",emailid)
    if res:
        print(f"{emailid} is valid")
        print(res)
    else:
        print(f"{emailid} is invalid")
except Exception as e:
    print("There is no matched pattern:",e)
   
```
