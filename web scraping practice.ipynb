```python
import requests
from bs4 import BeautifulSoup as bs

search_ele = input("Enter product name: ").replace(" ", "+")
url = "https://www.flipkart.com/search?q=" + search_ele

try:
    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept": "text/html,application/xhtml+xml",
        "Accept-Language": "en-US,en;q=0.9"
    }

    content = requests.get(url, headers=headers)

    print(content)

    web_page = bs(content.text, "html.parser")

    # Products
    products = web_page.find_all("a", class_="k7wcnx")

    print("Total Products:", len(products))
    print("=" * 60)

    product_data = []

    for product in products:

        # Product Link
        try:
            product_link = "https://www.flipkart.com" + product.get("href")
        except:
            product_link = "No product link"

        # Product Name
        try:
            product_name = product.find(
                "div", class_="RG5Slk"
            ).text.strip()
        except:
            product_name = "No product name"

        # Rating
        try:
            rating = product.find(
                "div", class_="MKiFS6"
            ).text.strip()
        except:
            rating = "No rating"

        # Discount Price / Current Price
        try:
            price = product.find(
                "div", class_="hZ3P6w DeU9vF"
            ).text.strip()
        except:
            price = "No price"

        # Actual Price
        try:
            actual_price = product.find(
                "div", class_="kRYCnD gxR4EY"
            ).text.strip()
        except:
            actual_price = "No actual price"

        # Product Description
        try:
            product_desc = []

            description = product.find_all(
                "li", class_="DTBslk"
            )

            for desc in description:
                product_desc.append(desc.text.strip())

        except:
            product_desc = ["No product description"]

        # Product Information
        product_info = {
            "Product Name": product_name,
            "Actual Price": actual_price,
            "Discount Price": price,
            "Rating": rating,
            "Product Description": product_desc,
            "Product Link": product_link
        }

        product_data.append(product_info)

    # Print all products
    for i, product in enumerate(product_data, start=1):

        print("Product:", i)
        print("Product Name:", product["Product Name"])
        print("Actual Price:", product["Actual Price"])
        print("Discount Price:", product["Discount Price"])
        print("Rating:", product["Rating"])
        print("Product Description:", product["Product Description"])
        print("Product Link:", product["Product Link"])
        print("-" * 60)

    print("Total Products Scraped:", len(product_data))

except Exception as e:
    print("Not able to scrap website:", e)

finally:
    print("valid")
```
